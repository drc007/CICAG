# Visual Studio Code (VS Code) Overview

### What is VS Code?

[Visual Studio Code](https://code.visualstudio.com/) is a free, lightweight, and powerful Integrated Development Environment (IDE) developed by Microsoft.

It supports many programming languages including JavaScript, C++, Java, and importantly for this course Python more through built-in features and extensions.

---

## What is an Integrated Development Environment (IDE)?

An IDE is a software application that provides tools for:

- Writing code (Editor)
- Running code
- Integrated terminal
- Debugging programs
- Managing files and projects
- Version control (Git)
- Installing extensions

VS Code combines all these tools into one interface.

---


# VS Code A brief introduction

As will be common throughout this course we will provide an overview of a tool and a concept - if you need more information the internet is your friend

The vertical bar on the left side of VS Code is called the **Activity Bar**.

Here is what each main icon means:

### 1. Explorer (📁)
- View and manage project files and folders
- Create new files
- Open existing files

### 2. Search (🔍)
- Search text across all project files
- Replace text in multiple files

### 3. Source Control (🔁)
- Git integration
- Track file changes
- Commit and push code to repositories

### 4. Run and Debug (🐞▶️)
- Run your program
- Debug code step-by-step
- View variables and breakpoints

### 5. Extensions (🧩)
- Install additional features
- For this course Jupyter, Python extensions will be used
- `Important note` Essential extensions should install when needed eg. loading a python file for the first time

## Opening a .ipynb File in VS Code

A `.ipynb` file is a Jupyter Notebook file.

1. Open VS Code
2. Click the Explorer icon (📁)
3. Open a folder
4. Click on a `.ipynb` file

The file will open in Notebook view with:
- Code cells
- Markdown cells
- Run buttons

You can execute cells using the ▶ Run button.
