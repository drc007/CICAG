# Miniforge Overview and enviroment setup

## What is Miniforge?

Miniforge is a minimal installer for Conda, a package and environment manager 

This doesn't require a license - other options for personal use include anaconda

It is community-driven and maintained under the conda-forge project.

Miniforge allows you to:
- Install Python and other languages
- Manage packages
- Create isolated environments
- Avoid dependency conflicts

---

## What is a Package Manager?

A package manager is a tool that:
- Downloads software packages
- Installs them automatically
- Manages dependencies
- Updates and removes packages safely

Conda (installed via Miniforge) is both:
- A package manager
- An environment manager


## Downloading Miniforge

1. Go to the official Miniforge GitHub releases page. - download here ([text](https://conda-forge.org/download/))
2. Download the installer for your operating system:
   - Windows (.exe)
   - macOS (.pkg)
   - Linux (.sh)
3. Run the installer.
4. Allow it to initialize Conda.

After installation, open a new terminal and verify:

`conda --version`

If Conda is installed correctly, it will print the version number.


# Create an env

Create a new Conda environment called CICAG_env with Python 3.10

`conda create -n CICAG_env python=3.10`

 `conda`

This is the **Conda package and environment manager**.

Conda is used to:

* Create environments
* Install packages
* Manage dependencies
* Control Python versions

It comes installed with **Miniforge**, **Anaconda**, or **Miniconda**.


 `create`

This tells Conda to:

> Create a new environment.

An environment is an isolated workspace that contains:

* Its own Python version
* Its own installed libraries
* No interference with other projects


 `-n`

This flag means:

> Name the environment.

It is the short version of:

```bash
--name
```


`CICAG_env`

This is the name of the environment you are creating.

You can name environments anything, for example:

* `data_science`
* `rdkit_env`
* `project1`
* `CICAG_env`

After creating it, you activate it using:

```bash
conda activate CICAG_env
```

Notice how `(CICAG_env)` in brackets


Finally install a package - here we install numpy (numerical python)

`conda install -n CICAG_env numpy`
