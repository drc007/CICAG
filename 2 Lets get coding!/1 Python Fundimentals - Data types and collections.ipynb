# The fundimentals and good coding practice

Python is an **Object-Oriented Programming (OOP)** language. In chemistry terms, think of an "Object" as a **Stock Solution**. 

A stock solution isn't just a liquid; it has **properties** (concentration, pH, temperature) and **behaviors** (it can be diluted, reacted, or titrated). 

Just as a solution can contain a mixture of different compounds, a Python object can contain a mixture of different **data types**.  

In coding language an object is data and its associated functionally that can be stored in memory.

An **object** has:
- A **type** (what kind of thing it is)
- A **value** (the data it holds)
- **Methods** (actions it can perform)

Think of it like an element: Carbon has a type (nonmetal), a value (atomic number 6), and properties/behaviours (forms 4 bonds, makes CO₂, etc.)

### Data Structures vs. Data Types

#### The 4 Primitive Data Types 
These are the simplest forms of data that cannot be broken down further:
- **Integer**: Whole numbers (e.g., number of protons: `6`)
- **Float**: Decimals (e.g., molar mass: `12.011`)
- **String**: Text (e.g., element symbol: `'C'`)
- **Boolean**: Logic (e.g., is the reaction exothermic? `True`)

| Type | Example | Description |
|------|---------|-------------|
| `int` | `6` | Whole numbers |
| `float` | `12.011` | Decimal numbers |
| `str` | `"Carbon"` | Text |
| `bool` | `True` / `False` | True or False |

#### Assignment
In python the `=` operator is assignment. This allows you to assign an object (eg. string, list) to a variable.

#### Strings in depth
Strings are particularly rich objects with lots of useful methods

```python
formula = "H2O"
print(len(formula))             # length
print(formula.startswith("H"))  # does it start with H?
print(formula.replace("H", "D")) # heavy water!


```

#### f-strings

Allow you to embed variables directly in text

```python
element = "Oxygen"
mass = 15.999
print(f"The atomic mass of {element} is {mass} g/mol")
```

#### The 4 Python Collections
These store collections of objects:
- **List**: An ordered sequence (e.g., a list of titration readings).
- **Tuple**: An unchangeable sequence (e.g., $(x, y, z)$ crystal coordinates).
- **Dictionary**: Key-value pairs (e.g., `{'Element': 'Gold', 'Symbol': 'Au'}`).
- **Set**: A collection of unique items (e.g., the set of distinct elements in a mixture).
