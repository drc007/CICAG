Python's official style guide is called **PEP 8**:  
https://peps.python.org/pep-0008/

It covers everything from spacing to naming. You don't need to memorise it - but you should understand its principles.

Let's look at the same block of code written two ways.


```python
# ❌ BAD - works, but unreadable
x= 2 
y=12.011
z=1
print((x*y)/z)

# ✅ GOOD - readable, self-documenting
moles=2
molar_mass=12.011
purity_fraction=1

mass_of_carbon = (moles * molar_mass) / purity_fraction
print(f"Mass: {mass_of_carbon:.3f} g")
```

## Variable & Function Naming Conventions

Python has clear, community-agreed naming rules. Following them makes your code instantly familiar to other Python developers.

| What | Convention | Example |
|------|-----------|--------|
| Variables | `lower_case` | `atomic_mass` |
| Functions | `lower_case` | `calculate_ph()` |
| Constants | `UPPER_CASE` | `AVOGADRO_CONSTANT` |
| Classes | `PascalCase` | `ChemicalElement` |

### Naming: be specific, but not absurdly long

There's a balance between being descriptive and being verbose:

```python
# ❌ Too vague
x = 310

# ❌ Too verbose
the_temperature_of_the_reaction_mixture_in_kelvin_at_standard_conditions = 310

# ✅ Just right
reaction_temperature_kelvin = 310

# ✅ Also fine when context is clear (e.g. inside a function about temperature)
temperature_k = 310
```

### Avoid these common naming mistakes:

What are Python built-ins? Why is it a bad idea to have them as variable names?

```python
# ❌ Don't use single letters - except loop counters i, j, k or math convention
# ❌ Dont shadow built-in names  
list = [1, 2, 3]    # BAD - 'list' is a Python built-in!
type = "metal"      # BAD - 'type' is a Python built-in!

# ✅ Use descriptive alternatives
element_list = [1, 2, 3]
element_type = "metal"

# ❌ Abbreviations that aren't universally known
atm_mss = 12.011    # atomic mass? atmosphere mass??

# ✅ Spell it out
atomic_mass = 12.011
```

## Comments & Docstrings

Comments explain the **why**, not the **what**. The code itself should explain what - if you need a comment to explain what a line does, consider rewriting the line instead.

**Two types:**
- `#` inline comments - brief notes on a single line
- `""" """` docstrings - formal documentation for functions, classes, modules

```python
# ❌ BAD comments - state the obvious
atomic_mass = 12.011    # set atomic mass to 12.011
x = x + 1              # add 1 to x
```

```python
# ✅ GOOD comments - explain the WHY or non-obvious context
atomic_mass = 12.011    # IUPAC 2021 standard atomic weight

# Kelvin required - ideal gas law breaks down with Celsius
temperature_celsius=25
temperature_k = temperature_celsius + 273.15

# Skip the first reading - sensor warm-up period causes drift
all_readings=[0,1,2,3,4,5]
clean_readings = all_readings[1:]
```

This is a function - we will be writing our own soon but its important to know how to document them

```python
# Docstrings - document your functions properly
def celsius_to_kelvin(temperature_celsius):
    """
    Convert a temperature from Celsius to Kelvin.

    Args:
        temperature_celsius (float): Temperature in degrees Celsius.

    Returns:
        float: Temperature in Kelvin.

    Raises:
        ValueError: If temperature is below absolute zero (-273.15°C).

    Example:
        >>> celsius_to_kelvin(100)
        373.15
    """
    return temperature_celsius + 273.15


print(celsius_to_kelvin(100))
help(celsius_to_kelvin)   # docstring is accessible via help()
```
