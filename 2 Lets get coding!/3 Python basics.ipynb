## Operators

### Arithmetic Operators


```python
# Molar mass of water: 2 H + 1 O
mass_H = 1.008
mass_O = 15.999

molar_mass_water = (2 * mass_H) + mass_O
print(f"Molar mass of H2O: {molar_mass_water:.3f} g/mol")

# Other operators
print(10 / 3)    # division → float
print(10 // 3)   # floor division → int
print(10 % 3)    # modulo (remainder)
print(2 ** 10)   # exponentiation (2 to the power 10)
```

### Comparison Operators

```python
boiling_point_water = 100  # are there any issues with this variable name?
current_temp = 85

print(current_temp == boiling_point_water)   # equal to
print(current_temp != boiling_point_water)   # not equal to
print(current_temp < boiling_point_water)    # less than
print(current_temp >= boiling_point_water)   # greater than or equal
```

### Logical Operators

```python
pH = 6.5
temperature = 37  # °C

is_acidic = pH < 7
is_body_temp = temperature == 37

print(is_acidic and is_body_temp)   # both must be True
print(is_acidic or is_body_temp)    # at least one must be True
print(not is_acidic)                 # inverts the boolean
```

##  Loops

Loops let you **repeat** actions - making code more flexible, readable and useable

### `for` loops

```python
# Loop over a list of elements
noble_gases = ["Helium", "Neon", "Argon", "Krypton", "Xenon", "Radon"]

for gas in noble_gases:
    print(f"{gas} is a noble gas.")
```

```python
# range() — repeat a fixed number of times
# Simulate temperature increase of 10°C each step
temp = 20
for step in range(5):
    temp += 10
    print(f"Step {step + 1}: Temperature = {temp}°C")
```

### `while` loops

```python
# Simulate a radioactive decay countdown (half-life)
# Starting with 1000 atoms print all halflife until the first under 100
atoms = 1000
half_life = 5730  # years
years = 0
periods = 0

#print(f"Start: {atoms} atoms")
while atoms > 100:
    atoms = atoms / 2
    periods += 1
    years += half_life
    print(f"After {years} years ({periods} half-lives): {atoms:.1f} atoms remaining")
```

### Loop control: `break` and `continue`

```python
# Find the first toxic element in a list and stop
elements = ["Hydrogen", "Helium", "Lithium", "Beryllium", "Boron", "Carbon"]
toxic = {"Beryllium", "Lead", "Mercury", "Arsenic"}

for element in elements:
    if element in toxic:
        print(f" Found toxic element: {element}. Stopping scan.")
        break
    print(f" {element} — safe")
```

## Data Structures

Python has four main built-in data structures:

| Structure | Syntax | Ordered | Mutable | Unique |

### Lists — ordered, changeable collections

Python indexing starts from zero.

Lists can store primitive data types, other collections

Even other lists to create nested lists

```python
# Periodic table — first 5 elements
elements = ["Hydrogen", "Helium", "Lithium", "Beryllium", "Boron"]

print(elements[0])       # indexing (0-based)
print(elements[-1])      # last element
print(elements[1:3])     # slicing

elements.append("Carbon")   # add to end
elements.insert(0, "Hydrogen")  # insert at index
print(elements)
print(f"List length: {len(elements)}")
```

### Sets — unordered, unique values

```python
# Elements present in two different compounds
ethanol = {"Carbon", "Carbon", "Hydrogen", "Hydrogen","Hydrogen","Hydrogen","Hydrogen","Oxygen" }
water = {"Hydrogen", "Oxygen"}
methane = {"Carbon", "Hydrogen"}

# Print ethanol set - what do you expect the output to be?
```

```python
# Sets have unique operators

print("Common elements (ethanol ∩ water):", ethanol & water)      # intersection
print("All elements (ethanol ∪ methane):", ethanol | methane)     # union
print("Only in ethanol (not water):", ethanol - water)            # difference
print("Is methane a subset of ethanol?", methane.issubset(ethanol))
```

### Dictionaries — key-value pairs

```python
# A mini periodic table as a dictionary
periodic_table = {
    "H":  {"name": "Hydrogen",  "mass": 1.008,  "group": 1},
    "He": {"name": "Helium",    "mass": 4.003,  "group": 18},
    "C":  {"name": "Carbon",    "mass": 12.011, "group": 14},
    "O":  {"name": "Oxygen",    "mass": 15.999, "group": 16},
    "Na": {"name": "Sodium",    "mass": 22.990, "group": 1},
}

# Access by key the information associated with carbon
print(periodic_table["C"])

# Access by key the mass of oxygen
print(periodic_table["O"]["mass"])

# Iterate through the periodic table a return the symbol and all elements
# hint periodic_table.items()
```

```python
# Adding and checking keys
periodic_table["N"] = {"name": "Nitrogen", "mass": 14.007, "group": 15}


# check if a value is in the periodic table
print("N" in periodic_table)   # True
print("Au" in periodic_table)  # False

# Safe access with .get() — returns None if key missing by default
print(periodic_table.get("Au", "Element not found"))
```
