## Functions good coding practice

**DRY = Don't Repeat Yourself.**

If you find yourself writing the same block of code more than once, it should be a function. Functions make code:
- Easier to test
- Easier to debug (fix in one place, not ten)
- Easier to read


```python
# ❌ WET code (Write Everything Twice) — repeated logic
mass_nacl = 5.0
moles_nacl = mass_nacl / 58.44
concentration_nacl = moles_nacl / 0.5

mass_hcl = 3.6
moles_hcl = mass_hcl / 36.46
concentration_hcl = moles_hcl / 0.5

mass_naoh = 8.0
moles_naoh = mass_naoh / 40.00
concentration_naoh = moles_naoh / 0.5

# What if the volume changes? You'd need to change it in 3 places!
```

```python
#answer

# ✅ DRY code — logic defined once, reused everywhere
def calculate_concentration(mass_g, molar_mass, volume_litres):
    """
    Calculate molar concentration of a solution.

    Args:
        mass_g (float): Mass of solute in grams.
        molar_mass (float): Molar mass of solute in g/mol.
        volume_litres (float): Volume of solution in litres.

    Returns:
        float: Concentration in mol/L (M).
    """
    moles = mass_g / molar_mass
    return moles / volume_litres


volume = 0.5  # litres — change here and it updates everywhere

conc_nacl = calculate_concentration(5.0, 58.44, volume)
conc_hcl  = calculate_concentration(3.6, 36.46, volume)
conc_naoh = calculate_concentration(8.0, 40.00, volume)

print(f"NaCl: {conc_nacl:.3f} M")
print(f"HCl:  {conc_hcl:.3f} M")
print(f"NaOH: {conc_naoh:.3f} M")
```
