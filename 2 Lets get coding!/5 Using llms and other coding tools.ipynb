#### How to write good prompts for coding

The quality of the output depends heavily on the quality of the input. Follow these principles:

**Be specific about context, inputs, outputs, and constraints.**

```python
# EXAMPLE PROMPTS — comparing vague vs specific

bad_prompt = """
Write code to calculate pH.
"""
# Problem: What input? What formula? What should it return? What edge cases?

good_prompt = """
Write a Python function called `calculate_ph` that:
- Takes a single argument: `hydrogen_ion_concentration` (a float, in mol/L)
- Returns the pH as a float, rounded to 2 decimal places
- Raises a ValueError if the concentration is <= 0
- Includes a docstring with an example

Use the formula: pH = -log10([H+])
"""
# Much better — the LLM knows exactly what to build

"Compare these two prompts — the second shoudl produce far better code!"
```

Here's what the LLM would (should) produce from the good prompt:

See the comments for follow up qustions?

```python

import math   # Should math if we already have numpy imported in our code?

def calculate_ph(hydrogen_ion_concentration):
    """
    Calculate the pH from hydrogen ion concentration.

    Args:
        hydrogen_ion_concentration (float): [H+] concentration in mol/L.

    Returns:
        float: pH value rounded to 2 decimal places.

    Raises:
        ValueError: If concentration is zero or negative.

    Example:
        >>> calculate_ph(1e-7)
        7.0
    """
    if hydrogen_ion_concentration <= 0:
        raise ValueError("Hydrogen ion concentration must be greater than 0.")   # Can you explain how error handeling works?
    return round(-math.log10(hydrogen_ion_concentration), 2)
```

### Critical warnings when using LLMs for code

LLMs are powerful but have real limitations. **Never blindly trust generated code.**

```python
# THINGS TO ALWAYS DO AFTER LLM GENERATES CODE:

# 1. READ IT — understand every line before using it
#    If you can't explain what a line does, you shouldn't use it.

# 2. TEST IT — run it with known inputs and check the output
def llm_generated_moles(mass, molar_mass):
    """Always test with known values!"""
    return mass / molar_mass

# Test with values you can verify by hand
assert llm_generated_moles(18.015, 18.015) == 1.0, "1 mole of water should be 18.015g"
assert llm_generated_moles(58.44, 58.44) == 1.0, "1 mole of NaCl should be 58.44g"
print("Tests passed ")

# 3. CHECK THE MATHS — for scientific code, verify formulas independently
# 4. CHECK FOR EDGE CASES — what happens with 0, negative numbers, empty lists?
# 5. QUESTION THE LIBRARY CHOICES — LLMs sometimes suggest deprecated functions
```

```python
# LLMs can hallucinate — invent functions that don't exist!
# EXAMPLE of what NOT to trust blindly:

hallucination_example = """
# An LLM might generate this looks plausible but is WRONG:

import chemistry_utils   # ← this package might not exist!

pH = chemistry_utils.calculate_ph_from_concentration(0.001)  # ← invented function!
"""
```

### LLMs as a learning tool, not a crutch

The best use of LLMs is to **accelerate your learning**, not bypass it.

```python
# BAD habit — paste in output, move on, never understand it
# This creates fragile code you can't debug or extend

# GOOD habit — use LLMs to explain concepts
learning_prompts = [
    "Explain what a list comprehension is and show 3 examples, from simple to complex.",
    "I wrote this code: [show code]. What would be a more Pythonic way to write it?",
    "What does `enumerate()` do? Show me an example using a list of chemical elements.",
    "What's the difference between a list and a generator? When should I use each?",
    "Review this function for best practices and suggest improvements: [paste function]",
]
```

#### The Golden Rule of LLM-assisted coding

> **"The LLM is a fast, knowledgeable colleague who sometimes hallucinates. You are the expert who verifies, understands, and takes responsibility for the code."**

| Do ✅ | Don't ❌ |
|-------|---------|
| Use it to scaffold and speed up | Use it as a black box |
| Ask it to explain what it wrote | Paste output straight into production |
| Test generated code thoroughly | Trust it for numerical/scientific accuracy |
| Ask follow-up questions | Accept first output without critique |
| Use it to learn new syntax | Rely on it as a substitute for understanding |
