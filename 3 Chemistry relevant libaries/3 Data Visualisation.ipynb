# Matplotlib — Visualising Chemistry

Matplotlib is Python's core plotting library. It gives you complete control over every element of a figure.

## Anatomy of a Matplotlib Figure

```
Figure  ← the whole canvas
  └── Axes  ← one plot / subplot (has x-axis, y-axis, title, etc.)
        ├── Title
        ├── X-axis label
        ├── Y-axis label
        ├── Data (lines, scatter, bars...)
        └── Legend
```

The two ways to use Matplotlib:
 1. plt.plot() — quick and simple ("pyplot interface")
 2. fig, ax = plt.subplots() — full control ("object-oriented interface")
 
Always prefer the object-oriented approach for anything non-trivial.

````python
import numpy as np
import matplotlib.pyplot as plt


# Simple example: ideal gas pressure vs volume at constant temperature
# PV = nRT → P = nRT/V
n = 1.0       # moles
R = 8.314     # J mol⁻¹ K⁻¹
T = 298       # K (25°C)

volumes = np.linspace(0.001, 0.050, 200)   # m³
pressures = (n * R * T) / volumes / 1000   # kPa

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(volumes * 1000, pressures, color="steelblue", linewidth=2.5, label="Ideal gas (298 K)")

ax.set_title("Ideal Gas: Pressure vs Volume (n = 1 mol, T = 298 K)", fontsize=14, fontweight="bold")
ax.set_xlabel("Volume (L)", fontsize=12)
ax.set_ylabel("Pressure (kPa)", fontsize=12)
ax.legend(fontsize=11)
ax.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
````
