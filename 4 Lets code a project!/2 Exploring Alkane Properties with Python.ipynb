
# Getting the Data

Our dataset contains the first 15 straight-chain alkanes (methane → pentadecane) with experimentally measured physical properties.

In a real project this would be a file on disk: `pd.read_csv("alkanes.csv")`  
Here we build it inline so nothing needs downloading.

**Columns:**
- `name` - IUPAC name
- `formula` - molecular formula
- `smiles` - SMILES string (input for RDKit)
- `carbon_count` - number of carbons (n)
- `boiling_point_c` - boiling point in °C at 1 atm
- `melting_point_c` - melting point in °C at 1 atm
- `density_g_per_ml` - liquid density at 20°C
- `vapour_pressure_kpa` - vapour pressure at 20°C

```python

# Load into a DataFrame
df = pd.read_csv(r'Exploring Alkane Properties with Python.csv') # whats the R here whats the need?

df["vapour_pressure_kpa"] = pd.to_numeric(df["vapour_pressure_kpa"], errors="coerce")
df["density_g_per_ml"]    = pd.to_numeric(df["density_g_per_ml"],    errors="coerce")

# Why do we use .to_numeric

df
```
